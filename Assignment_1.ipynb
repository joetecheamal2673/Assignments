```python
#1)	Using a for loop, print each job posting in a readable format.

job_data = {
    'Job_ID': [1, 2, 3, 4, 5],
    'Role': ['Data Analyst', 'Data Scientist', 'Business Analyst',
             'Data Engineer', 'ML Engineer'],
    'Company': ['ABC Corp', 'XyZ Ltd', 'DataWorks',
                'InfraTech', 'AI Labs'],
    'Skills': ['Excel SQL Python', 'Python ML Statistics',
               'Excel SQL PowerBI', 'Python SQL Spark',
               'Python ML DeepLearning']
}

for i in range(len(job_data["Job_ID"])):
    print(
        "ID:", job_data["Job_ID"][i],
        "| Role:", job_data["Role"][i],
        "| Company:", job_data["Company"][i],
        "| Skills:", job_data["Skills"][i]
    )
```

```python
#2).Add New Job Postings (interactive)

n = int(input("How many job postings do you want to add? "))

for i in range(n):
    new_id = max(job_data["Job_ID"]) + 1
    role = input("Enter role: ")
    company = input("Enter company: ")
    skills = input("Enter skills (space-separated or comma-separated): ")
    job_data["Job_ID"].append(int(new_id))
    job_data["Role"].append(role)
    job_data["Company"].append(company)
    job_data["Skills"].append(skills)

print("\nUpdated Job Data:")
print(job_data)
```

```python
#3)	Clean & Normalize Skills (text processing)

import re

def clean_skills(skills):
    skills = skills.lower()
    skills = skills.replace(",", " ")
    skills = re.sub(r'\s+', ' ', skills)
    skills = skills.strip()

    return skills
job_data['Skills'] = [
    clean_skills(skill) for skill in job_data['Skills']
]
print(job_data['Skills'])
```

```python
# 4).Skill Extraction & Frequency

all_skills = []

for skill_string in job_data['Skills']:
    skills = skill_string.split()
    all_skills.extend(skills)

skill_count = {}

for skill in all_skills:
    if skill in skill_count:
        skill_count[skill] += 1
    else:
        skill_count[skill] = 1

top_skills = sorted(
    skill_count.items(),
    key=lambda x: x[1],
    reverse=True
)

skill_names = {
    'python': 'Python',
    'sql': 'sql',
    'excel': 'Excel',
    'ml': 'ML',
    'powerbi': 'Power BI'
}
print("Top skills by frequency:\n")

for skill, count in top_skills[:5]:
    print(f"{skill_names.get(skill, skill)}: {count}")
```

```python
# 5).Basic Statistics & Insights

total_jobs = len(job_data['Job_ID'])

unique_skills = set()

for skill_string in job_data['Skills']:
    skills = skill_string.lower().split()
    unique_skills.update(skills)

python_jobs = []

for i in range(len(job_data['Skills'])):
    skills = job_data['Skills'][i].lower().split()

    if 'python' in skills:
        python_jobs.append(job_data['Role'][i])

print("Total Jobs:", total_jobs)
print("Unique skills:", len(unique_skills))
print("Jobs mentioning Python:", python_jobs)
```

```python
# 6).Skill Search Function (functions & conditional logic)

def find_jobs_by_skill(skill_query):


    skill_query = skill_query.lower().strip()

    matching_jobs = []


    if skill_query == "":
        print("Please enter a valid skill.")

    elif skill_query:

        for i in range(len(job_data['Skills'])):

            skills = job_data['Skills'][i].lower().split()

            if skill_query in skills:
                matching_jobs.append((
                    job_data['Job_ID'][i],
                    job_data['Role'][i],
                    job_data['Company'][i]
                ))

    else:
        print("No matching skill found.")


    return matching_jobs



result = find_jobs_by_skill('sql')

print("Jobs matching SQL:")

for job in result:
    print(job)
```

```python

# 7).Lambda Filtering & List Comprehension

roles_with_common_skills = []
roles_with_common_skills = list(filter(
    lambda role: any(
        skill_count[skill] >= 2
        for skill in job_data['Skills'][job_data['Role'].index(role)].split()
    ),
    job_data['Role']
))
print("Roles with at least one common skill:", roles_with_common_skills)
```

```python
# 8).Unique Skill Set

unique_skills = set(all_skills)

sorted_skills = sorted(unique_skills)

print("Unique skills:")
print(sorted_skills)
```

```python
# 9).Save Cleaned Data to File (file handling & exceptions)

import csv

try:
    with open('job_skills.csv', 'w', newline='') as file:
        writer = csv.writer(file)

        # Write header
        writer.writerow(['Job_ID', 'Role', 'Company', 'Skills'])

        # Write job records
        for i in range(len(job_data['Job_ID'])):
            writer.writerow([
                job_data['Job_ID'][i],
                job_data['Role'][i],
                job_data['Company'][i],
                job_data['Skills'][i]
            ])

except IOError as e:
    print("Error while creating CSV file:", e)

else:
    print("job_skills.csv created successfully!")

finally:
    print("CSV file operation completed.")

try:
    with open('top_skills.txt', 'w') as file:

        file.write("Top skills by frequency:\n\n")

        for skill, count in top_skills[:5]:
            file.write(f"{skill}: {count}\n")

except IOError as e:
    print("Error while creating text file:", e)

else:
    print("top_skills.txt created successfully!")

finally:
    print("Text file operation completed.")
```
