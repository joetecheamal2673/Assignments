```python

# SECTION 2 - Iteration, Logic & Reusable Functions

transactions = [45.0, 0.0, 120.5, -10.0, 300.0, 85.0]

for amount in transactions:

    if amount <= 0:
        continue

    if amount > 250.0:
        print("High-value order detected. Security review required!")
        break

    print("Valid transaction amount:", amount)


def calcite_discount(bill_amount, vip_member=False):

    if vip_member == True:
        discounted_total = bill_amount * 0.85

    else:
        discounted_total = bill_amount * 0.95

    return discounted_total

print("\nVIP Customer Bill:", calcite_discount(1000, True))
print("Regular Customer Bill:", calcite_discount(1000))

points_earned = lambda amount: amount // 10

print("Reward Points:", points_earned(250))

def tier_bonus(level):

    if level == 1:
        return 50

    else:
        return 50 + tier_bonus(level - 1)


print("Tier Bonus Points:", tier_bonus(3))
```

```python

#***Section5: Oblect-Oriented Programming(OOP) Architecture

class StoreBranch:

    company_name = "Urban Retail"

    def __init__(self, branch_id, daily_revenue):

        self.branch_id = branch_id

        self.__daily_revenue = daily_revenue

    def get_revenue(self):

        return self.__daily_revenue


class FlagshipStore(StoreBranch):

    def __init__(self, branch_id, daily_revenue, vip_lounge_access):

        super().__init__(branch_id, daily_revenue)

        self.vip_lounge_access = vip_lounge_access

    def get_revenue(self):

        return super().get_revenue() + 500.0


flagship = FlagshipStore(101, 2000.0, True)


print("Company Name:", flagship.company_name)
print("Branch ID:", flagship.branch_id)
print("VIP Lounge Access:", flagship.vip_lounge_access)


print("Final Calculated Revenue:", flagship.get_revenue())
```
