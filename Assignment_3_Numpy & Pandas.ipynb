```python
#2.	Inspection and Properties -
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Shape of the array:", temperatures_w1.shape)
print("Data type of the array:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
#3.Array Operations -
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_f = (temperatures_w1 * 9/5) + 32
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("Temperatures in Celsius:", temperatures_w1)
print("Temperatures in Fahrenheit:", temperatures_f)
print("Maximum Temperature:", max_temp, "°C")
print("Minimum Temperature:", min_temp, "°C")
print("Mean Temperature:", mean_temp, "°C")
```

```python
#4.	Array Slicing and Indexing -
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
first_three_days = temperatures_w1[:3]
weekend_temperatures = temperatures_w1[-2:]
middle_three_days = temperatures_w1[2:5]


print("First three days:", first_three_days)
print("Weekend temperatures:", weekend_temperatures)
print("Middle three days:", middle_three_days)
```

```python
#1.	Creating Pandas Series -

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
)

print(marks)
```

```python
#2.Indexing and Slicing -

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
)

first_rank_mark = marks.iloc[0]
print("1st Rank Student Mark:", first_rank_mark)

top_3_ranks = marks.loc[["Rank1", "Rank2", "Rank3"]]
print("\nMarks of Top 3 Ranks:")
print(top_3_ranks)

third_rank_mark = marks.iloc[2]
print("\n3rd Rank Student Mark:", third_rank_mark)

high_marks = marks[marks > 90]
print("\nRanks with Marks Greater than 90:")
print(high_marks)
```

```python
#3.Manipulating Series -

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
)

marks.loc["Rank1"] = 100
print("Updated Marks:")
print(marks)

marks = marks.drop("Rank5")
print("\nMarks after removing the last rank:")
print(marks)

cgpa = marks / 10
print("\nCGPA:")
print(cgpa)
```

```python
#2.Data Exploration-

import pandas as pd

transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    "ProductCategory": [
        "Electronics", "Clothing", "Electronics", "Furniture", "Clothing",
        "Electronics", "Furniture", "Clothing", "Furniture", "Electronics"
    ],

    "Region": [
        "North", "South", "North", "East", "West",
        "North", "East", "West", "South", "North"
    ],

    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})


print("Complete Transactions DataFrame:")
print(transactions)
print("\nHead of the DataFrame:")
print(transactions.head())
print("\nTail of the DataFrame:")
print(transactions.tail())
print("\nShape of the DataFrame:")
print(transactions.shape)
print("\nColumn Names:")
print(transactions.columns)
print("\nData Types:")
print(transactions.dtypes)
print("\nProductCategory and Amount:")
print(transactions[["ProductCategory", "Amount"]])
print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])


filtered_data = transactions[
    (transactions["Region"] == "North") &
    (transactions["Amount"] > 200)
]

print("\nTransactions in North with Amount greater than 200:")
print(filtered_data)
print("\nProduct Category Value Counts:")
print(transactions["ProductCategory"].value_counts())
print("\nUnique Regions:")
print(transactions["Region"].unique())

region_mean = transactions.groupby("Region")["Amount"].mean()

print("\nMean Amount for Each Region:")
print(region_mean)
```

```python
#3.Manipulating the Data frame-

import pandas as pd

transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": [
        "Electronics", "Clothing", "Electronics", "Furniture", "Clothing",
        "Electronics", "Furniture", "Clothing", "Furniture", "Electronics"
    ],
    "Region": [
        "North", "South", "North", "East", "West",
        "North", "East", "West", "South", "North"
    ],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165

print("After modifying TransactionID 102:")
print(transactions)

transactions["Discount"] = transactions["Amount"] * 0.10

print("\nAfter adding Discount column:")
print(transactions)

transactions = transactions[
    transactions["TransactionID"] != 109
]

print("\nAfter removing TransactionID 109:")
print(transactions)

transactions.drop(columns=["Discount"], inplace=True)

print("\nFinal DataFrame after deleting Discount column:")
print(transactions)
```
